```python
import streamlit as st
import pandas as pd
```

```python
# サンプルデータ
data = [
    ["1月", "日本", 100],
    ["1月", "中国", 200],
    ["1月", "アメリカ", 150],
    ["2月", "中国", 150],
    ["2月", "アメリカ", 180],
    ["3月", "日本", 120],
    ["3月", "中国", 160],
]

df = pd.DataFrame(data, columns=["month", "country", "export"])

# 表形式に変換
table = df.pivot(index="month", columns="country", values="export").fillna(0)

st.title("国別輸出量")

# 国リスト
countries = table.columns.tolist()

# 国選択（複数選択）
selected = st.multiselect(
    "国を選択してください",
    countries,
    default=countries[:1]
)

# 選択された国のデータ
if selected:
    st.line_chart(table[selected])
else:
    st.write("国を選択してください")
```
